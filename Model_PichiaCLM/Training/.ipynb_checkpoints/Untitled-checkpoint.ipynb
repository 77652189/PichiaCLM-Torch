```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

import numpy as np
import pandas as pd
import pickle
import os

# ---------------------------------------------------------
# 1. Load Data and Settings (Same as your original)
# ---------------------------------------------------------
with open("Pichia_2Target.pkl", "rb") as fp:
    Data_AllOrg = pickle.load(fp)
    
AA_tr = Data_AllOrg['AA_tr']
Cds_tr = Data_AllOrg['Cds_tr']
# AA_ts = Data_AllOrg['AA_ts']
# Cds_ts = Data_AllOrg['Cds_ts']

Settings = pd.read_csv('./BO_forHyperParameter/Arch1/Round3.csv').iloc[:, 1:]
Setting_no = 1

# Hyperparameters
Max_length = 1000
learning_rate = 0.001
batch_size = 150
epochs = 100
aa_vocab_size = 25
dna_vocab_size = 67

hidden_size_enc = int(Settings['Enc hidden size'][Setting_no])
hidden_size_enc_aa = int(Settings['Enc hidden size'][Setting_no])
embedding_size_enc = int(Settings['Enc Embedding size'][Setting_no])
embedding_size_dec = int(Settings['Dec Embedding size'][Setting_no])
Dense_layer_size = int(Settings['Dense Layer size'][Setting_no])
Dense_layer_size_aa = int(Settings['Dense Layer size aa'][Setting_no])

drop_rate = float(Settings['Drop rate'][Setting_no])
drop_rate_aa = float(Settings['Drop rate aa'][Setting_no])

# Setup Device (GPU if available)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ---------------------------------------------------------
# 2. PyTorch Model Definition
# ---------------------------------------------------------
class DotProductAttention(nn.Module):
    def forward(self, query, value):
        # query: (batch, seq_q, hidden)
        # value (encoder outputs): (batch, seq_v, hidden)
        # Compute attention scores
        scores = torch.bmm(query, value.transpose(1, 2))
        weights = F.softmax(scores, dim=-1)
        # Context vector
        context = torch.bmm(weights, value)
        return context

class EncDecModel(nn.Module):
    def __init__(self):
        super(EncDecModel, self).__init__()
        
        # Embeddings (padding_idx=0 acts as Keras mask_zero=True)
        self.encod_emb = nn.Embedding(num_embeddings=aa_vocab_size, embedding_dim=embedding_size_enc, padding_idx=0)
        self.dex = nn.Embedding(num_embeddings=dna_vocab_size, embedding_dim=embedding_size_dec, padding_idx=0)
        
        # Encoder
        self.encoder = nn.GRU(input_size=embedding_size_enc, hidden_size=hidden_size_enc, 
                              batch_first=True, bidirectional=True)
        
        # Decoders
        self.decoder = nn.GRU(input_size=embedding_size_dec, hidden_size=2*hidden_size_enc, batch_first=True)
        self.decoder_aa = nn.GRU(input_size=embedding_size_enc, hidden_size=2*hidden_size_enc_aa, batch_first=True)
        
        # Attention Layers
        self.attn = DotProductAttention()
        self.attn_aa = DotProductAttention()
        
        # Intermediate layers (TimeDistributed is inherently handled by PyTorch Linear layers on 3D tensors)
        # Input to linear will be: concat(decoder_sequence, attn_out) -> 2 * 2*hidden_size_enc = 4*hidden_size_enc
        self.intermediate = nn.Sequential(
            nn.Linear(4 * hidden_size_enc, Dense_layer_size),
            nn.Tanh(),
            nn.Dropout(drop_rate)
        )
        
        self.intermediate_aa = nn.Sequential(
            nn.Linear(4 * hidden_size_enc_aa, Dense_layer_size_aa),
            nn.Tanh(),
            nn.Dropout(drop_rate_aa)
        )
        
        # Final output layers (No softmax here, PyTorch CrossEntropyLoss expects raw logits)
        self.out_dna = nn.Linear(Dense_layer_size, dna_vocab_size)
        self.out_aa = nn.Linear(Dense_layer_size_aa, aa_vocab_size)

    def forward(self, input_sequence, decoder_inputs, decoder_inputs_aa):
        # --- Encoder ---
        embedding = self.encod_emb(input_sequence)
        encoder_sequence, encoder_final = self.encoder(embedding)
        
        # Bidirectional GRU returns (num_layers * num_directions, batch, hidden)
        # We need to concatenate the forward and backward final states: shape (1, batch, 2*hidden)
        enc_final_f = encoder_final[0, :, :]
        enc_final_b = encoder_final[1, :, :]
        encoder_final_concat = torch.cat([enc_final_f, enc_final_b], dim=-1).unsqueeze(0)
        
        # --- Decoder ---
        final_dex = self.dex(decoder_inputs)
        final_dex_aa = self.encod_emb(decoder_inputs_aa)
        
        decoder_sequence, _ = self.decoder(final_dex, encoder_final_concat)
        decoder_sequence_aa, _ = self.decoder_aa(final_dex_aa, encoder_final_concat)
        
        # --- Attention ---
        attn_out = self.attn(decoder_sequence, encoder_sequence)
        attn_out_aa = self.attn_aa(decoder_sequence_aa, encoder_sequence)
        
        # --- Concatenation ---
        decoder_concat_input = torch.cat([decoder_sequence, attn_out], dim=-1)
        decoder_concat_input_aa = torch.cat([decoder_sequence_aa, attn_out_aa], dim=-1)
        
        # --- Intermediate & Output ---
        inter_out = self.intermediate(decoder_concat_input)
        inter_out_aa = self.intermediate_aa(decoder_concat_input_aa)
        
        logits = self.out_dna(inter_out)
        logits_aa = self.out_aa(inter_out_aa)
        
        return logits, logits_aa

# ---------------------------------------------------------
# 3. Data Preparation
# ---------------------------------------------------------
# Slicing matching your Keras fit() method
X_enc = AA_tr[:, 1:Max_length+1]
X_dec_dna = Cds_tr[:, 0:Max_length-1]
X_dec_aa = AA_tr[:, 0:Max_length]

Y_dna = Cds_tr[:, 1:Max_length]
Y_aa = AA_tr[:, 1:Max_length+1]

# Convert arrays to PyTorch Tensors
X_enc_t = torch.tensor(X_enc, dtype=torch.long)
X_dec_dna_t = torch.tensor(X_dec_dna, dtype=torch.long)
X_dec_aa_t = torch.tensor(X_dec_aa, dtype=torch.long)
Y_dna_t = torch.tensor(Y_dna, dtype=torch.long)
Y_aa_t = torch.tensor(Y_aa, dtype=torch.long)

# Create Dataset and Validation Split (using 20% for val)
dataset = TensorDataset(X_enc_t, X_dec_dna_t, X_dec_aa_t, Y_dna_t, Y_aa_t)

val_size = int(0.2 * len(dataset))
train_size = len(dataset) - val_size
train_dataset, val_dataset = torch.utils.data.random_split(dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)

# ---------------------------------------------------------
# 4. Training Initialization
# ---------------------------------------------------------
model = EncDecModel().to(device)

# ignore_index=0 ensures that padding tokens do not contribute to the loss (matching mask_zero=True)
criterion = nn.CrossEntropyLoss(ignore_index=0)
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# Early stopping variables
best_val_loss = float('inf')
patience = 3
patience_counter = 0

checkpoint_path = "./PichiaData/2Target_AllData/FinArch1_AttnCorr_cp.weights.pth"
os.makedirs(os.path.dirname(checkpoint_path), exist_ok=True)

# ---------------------------------------------------------
# 5. Custom Training Loop
# ---------------------------------------------------------
print("Starting Training...")
for epoch in range(epochs):
    model.train()
    total_train_loss = 0
    
    for batch in train_loader:
        # Move batch to GPU/CPU
        b_x_enc, b_x_dec_dna, b_x_dec_aa, b_y_dna, b_y_aa = [t.to(device) for t in batch]
        
        optimizer.zero_grad()
        
        # Forward pass
        logits, logits_aa = model(b_x_enc, b_x_dec_dna, b_x_dec_aa)
        
        # PyTorch CrossEntropy expects shape (Batch, Classes, Seq_length)
        logits = logits.permute(0, 2, 1)
        logits_aa = logits_aa.permute(0, 2, 1)
        
        # Calculate loss for both targets
        loss_dna = criterion(logits, b_y_dna)
        loss_aa = criterion(logits_aa, b_y_aa)
        loss = loss_dna + loss_aa
        
        # Backward and optimize
        loss.backward()
        optimizer.step()
        
        total_train_loss += loss.item()
        
    avg_train_loss = total_train_loss / len(train_loader)
    
    # Validation Phase
    model.eval()
    total_val_loss = 0
    with torch.no_grad():
        for batch in val_loader:
            b_x_enc, b_x_dec_dna, b_x_dec_aa, b_y_dna, b_y_aa = [t.to(device) for t in batch]
            
            logits, logits_aa = model(b_x_enc, b_x_dec_dna, b_x_dec_aa)
            
            logits = logits.permute(0, 2, 1)
            logits_aa = logits_aa.permute(0, 2, 1)
            
            val_loss_dna = criterion(logits, b_y_dna)
            val_loss_aa = criterion(logits_aa, b_y_aa)
            total_val_loss += (val_loss_dna + val_loss_aa).item()
            
    avg_val_loss = total_val_loss / len(val_loader)
    
    print(f"Epoch {epoch+1}/{epochs} | Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f}")
    
    # Early Stopping & Checkpointing
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        patience_counter = 0
        torch.save(model.state_dict(), checkpoint_path)
        print(f"  -> Validation loss improved. Model saved to {checkpoint_path}")
    else:
        patience_counter += 1
        if patience_counter >= patience:
            print(f"Early stopping triggered after {epoch+1} epochs.")
            break
```
