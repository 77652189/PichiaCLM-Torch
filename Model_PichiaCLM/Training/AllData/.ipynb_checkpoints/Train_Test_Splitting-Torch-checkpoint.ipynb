```python
import torch
import pandas as pd
import numpy as np


def build_vocabularies():
    """构建氨基酸和密码子的字典映射，保持与原始代码逻辑完全一致"""
    
    # 1. 氨基酸字典 (严格按照原始映射)
    aa_vocab = {
        'A': 1, 'C': 2, 'D': 3, 'E': 4, 'F': 5, 'G': 6, 'H': 7, 'I': 8, 'K': 9, 'L': 10,
        'M': 11, 'N': 12, 'P': 13, 'Q': 14, 'R': 15, 'S': 16, 'T': 17, 'V': 18, 'W': 19,
        'Y': 20, 'X': 21, 'Z': 21, 'B': 21, 'U': 21, 'O': 21, '*': 22
    }
    
    # 2. 密码子字典 (按照原始 dic_AA_codon 提取出的顺序)
    dic_AA_codon = {
        'A': ['GCT', 'GCC', 'GCA', 'GCG'], 'C': ['TGT', 'TGC'],
        'D': ['GAT', 'GAC'], 'E': ['GAA', 'GAG'],
        'F': ['TTT', 'TTC'], 'G': ['GGT', 'GGA', 'GGC', 'GGG'],
        'H': ['CAT', 'CAC'], 'I': ['ATT', 'ATC', 'ATA'],
        'K': ['AAA', 'AAG'], 'L': ['TTA', 'TTG', 'CTT', 'CTC', 'CTA', 'CTG'],
        'M': ['ATG'], 'N': ['AAT', 'AAC'],
        'P': ['CCT', 'CCC', 'CCA', 'CCG'], 'Q': ['CAA', 'CAG'],
        'R': ['CGT', 'CGC', 'CGA', 'CGG', 'AGA', 'AGG'],
        'S': ['TCT', 'TCC', 'TCA', 'TCG', 'AGT', 'AGC'],
        'T': ['ACT', 'ACC', 'ACA', 'ACG'], 'V': ['GTT', 'GTC', 'GTA', 'GTG'],
        'W': ['TGG'], 'Y': ['TAT', 'TAC'],
        '*': ['TAA', 'TAG', 'TGA']
    }
    
    codon_vocab = {}
    idx = 1
    for aa in dic_AA_codon:
        for codon in dic_AA_codon[aa]:
            codon_vocab[codon] = idx
            idx += 1
            
    return aa_vocab, codon_vocab

def tokenize_sequences(aa_seqs, cds_seqs, aa_vocab, codon_vocab):
    """将文本序列转化为整数 Token 列表，使用原始代码指定的特殊 Token"""
    tokenized_aa = []
    tokenized_cds = []

    # 原始代码中的特殊 Token 设定
    AA_SOS_IDX = 24
    AA_EOS_IDX = 23
    AA_UNK_IDX = 21  # 对应原代码中的未知/特殊氨基酸 (X, Z, B, U, O)
    
    CDS_SOS_IDX = 65
    CDS_EOS_IDX = 66
    CDS_UNK_IDX = 0  # 原代码无未知密码子处理，这里用 0 兜底防报错

    for aa, cds in zip(aa_seqs, cds_seqs):
        # 编码氨基酸序列
        aa_tok = [AA_SOS_IDX] + [aa_vocab.get(c, AA_UNK_IDX) for c in aa] + [AA_EOS_IDX]
        tokenized_aa.append(aa_tok)
        
        # 编码密码子序列 (按3个字母切分)
        cds_tok = [CDS_SOS_IDX]
        for i in range(0, len(cds), 3):
            codon = cds[i:i+3]
            if len(codon) == 3:  # 确保是完整的密码子
                cds_tok.append(codon_vocab.get(codon, CDS_UNK_IDX))
        cds_tok.append(CDS_EOS_IDX)
        tokenized_cds.append(cds_tok)

    return tokenized_aa, tokenized_cds

def pad_and_truncate(sequences, maxlen, pad_idx=0):
    """对序列进行填充或截断，并确保截断后仍保留结束符"""
    padded = []
    for seq in sequences:
        if len(seq) > maxlen:
            # 截断超长序列，但保留最后一个元素 (EOS_IDX)
            seq = seq[:maxlen-1] + [seq[-1]]
        else:
            # 使用 PAD_IDX 填充不足长度的序列
            seq = seq + [pad_idx] * (maxlen - len(seq))
        padded.append(seq)
    
    # 转换为 PyTorch 的 LongTensor
    return torch.tensor(padded, dtype=torch.long)

def data_prep_pytorch(data_dict):
    """主数据处理管道"""
    print("开始清洗数据...")
    all_dfs = []
    # 1. 使用 Pandas 矢量化操作高效清洗数据
    for o, df in data_dict.items():
        # 剔除包含 'N' 的行 (使用 astype(str) 防止浮点数NaN报错)
        clean_df = df[~df['CDS_Seq'].astype(str).str.contains('N')]
        all_dfs.append(clean_df)
    
    # 合并所有生物体的数据
    combined_df = pd.concat(all_dfs, ignore_index=True)
    aa_seqs = combined_df['AA_Seq'].tolist()
    cds_seqs = combined_df['CDS_Seq'].tolist()
    
    # 2. 获取词表
    aa_vocab, codon_vocab = build_vocabularies()
    
    print(f"数据清洗完毕，共提取 {len(aa_seqs)} 条有效序列。正在分词...")
    # 3. 分词 (Tokenization)
    tokenized_aa, tokenized_cds = tokenize_sequences(aa_seqs, cds_seqs, aa_vocab, codon_vocab)
    
    # 4. 计算 95% 分位数长度 (统一 maxlen)
    # 因为 1 个氨基酸对应 1 个密码子，所以它们的 token 长度应该是一致的
    lengths = [len(seq) for seq in tokenized_aa]
    maxlen = int(np.percentile(lengths, 95))
    print(f"95% 分位数的序列长度为: {maxlen} (将以此长度进行统一填充/截断)")
    
    # 5. 填充并转换为 PyTorch 张量
    pad_idx = 0
    AA_tr = pad_and_truncate(tokenized_aa, maxlen, pad_idx)
    Cds_tr = pad_and_truncate(tokenized_cds, maxlen, pad_idx)
    
    print(f"数据准备完成。张量维度: AA_tr {AA_tr.shape}, Cds_tr {Cds_tr.shape}")
    return AA_tr, Cds_tr
```

```python
# ==========================================
# 使用示例
# ==========================================
if __name__ == '__main__':
    # 假设你已经有了 CSV 文件
    Data_dict = {
        0: pd.read_csv('CBS7435.csv'),  
        1: pd.read_csv('GS115_ext.csv'), 
        2: pd.read_csv('Kpastoris_WT.csv'),  
        3: pd.read_csv('Kphaffi_GS115_int.csv'),  
        4: pd.read_csv('Kphaffi_WT.csv')
    }

    # 获取处理好的 PyTorch Tensor
    AA_tr, Cds_tr = data_prep_pytorch(Data_dict)

    # 保存为 PyTorch 专用的 .pt 文件（比 pickle 更适合张量）
    torch.save({
        'AA_tr': AA_tr,
        'Cds_tr': Cds_tr
    }, 'Pichia_2Target.pt')
```
